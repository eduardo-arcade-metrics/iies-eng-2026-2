O programa percorre os 5 trechos da rota de um caminhão. Em cada trecho ele descobre o consumo (L/100 km) a partir da velocidade média, calcula os litros gastos e o tempo de viagem, e vai acumulando os totais. No fim calcula o custo do diesel e imprime um relatório resumido da rota.

```python
# Dados de entrada
extensoes_km = [42, 18, 65, 30, 25]      # distância de cada trecho (km)
velocidades_kmh = [90, 40, 110, 60, 80]  # velocidade média de cada trecho (km/h)
preco_litro = 6.89                       # preço do litro do diesel (R$)

# Acumuladores
distancia_total = 0.0   # distância total percorrida (km)
consumo_total = 0.0     # consumo total de combustível (L)
tempo_total = 0.0       # tempo total de viagem (h)
```

```python
maior_consumo_trecho = 0   # número do trecho que mais gastou litros
maior_consumo_litros = 0.0 # litros gastos nesse trecho

for i in range(len(extensoes_km)):
    extensao = extensoes_km[i]
    velocidade = velocidades_kmh[i]

    # Consumo (L/100km) conforme a velocidade média do trecho
    if velocidade <= 50:
        consumo_100km = 18.0
    elif velocidade <= 80:
        consumo_100km = 14.5
    elif velocidade <= 100:
        consumo_100km = 12.0
    else:
        consumo_100km = 15.5

    consumo_trecho = (extensao * consumo_100km) / 100  # litros gastos no trecho
    tempo_trecho_h = extensao / velocidade             # horas no trecho

    print(f"Trecho {i + 1}: {extensao} km a {velocidade} km/h | "
          f"Consumo: {consumo_100km:.1f} L/100km | "
          f"Gasto: {consumo_trecho:.2f} L | Tempo: {tempo_trecho_h:.2f} h")

    # Acumula os totais
    distancia_total += extensao
    consumo_total += consumo_trecho
    tempo_total += tempo_trecho_h

    if consumo_trecho > maior_consumo_litros:
        maior_consumo_litros = consumo_trecho
        maior_consumo_trecho = i + 1
```

```python
# Custo total = litros consumidos * preço do litro
custo_total = consumo_total * preco_litro
```

```python
print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total:.2f} km")
print(f"Tempo total:         {tempo_total:.2f} h")
print(f"Consumo total:       {consumo_total:.2f} litros")
print(f"Custo total:         R$ {custo_total:.2f}")
print("=========================")
print(f"Trecho com maior gasto em litros: {maior_consumo_trecho} ({maior_consumo_litros:.2f} L)")
```

## Interpretação do resultado

O trecho mais caro em litros foi o **Trecho 3** (65 km a 110 km/h), com cerca de 10,07 L. Isso acontece por dois motivos somados: ele é o trecho mais longo e foi percorrido acima de 100 km/h, faixa em que o consumo sobe para 15,5 L/100 km. Já o pior consumo *por quilômetro* foi o Trecho 2 (18,0 L/100 km, a 40 km/h), mas como ele tem só 18 km o gasto total foi pequeno.

A maior velocidade nem sempre é a mais econômica porque o arrasto aerodinâmico cresce com o quadrado da velocidade: acima de certo ponto o motor precisa de muito mais potência só para vencer o ar. Em velocidades muito baixas o problema é o oposto, o motor trabalha em regime ineficiente (marcha lenta, arranques e paradas). Por isso existe uma faixa intermediária (aqui, 81 a 100 km/h, com 12,0 L/100 km) que é a mais econômica, e o planejamento de rota pode aproveitar isso.
