Este programa calcula a tensão mecânica que uma carga provoca em um cabo de aço a partir do diâmetro do cabo e da força aplicada. Depois compara essa tensão com a tensão de ruptura do material (fator de utilização) e classifica o nível de risco da operação usando `if / elif / else`.

```python
# Dados de entrada
diametro_mm = 12          # diâmetro do cabo (mm)
carga_N = 8500            # força aplicada no cabo (N)
tensao_ruptura_MPa = 400  # tensão de ruptura do aço (MPa)
```

```python
import math

# Conversão de mm para m
diametro_m = diametro_mm / 1000

# Área da seção transversal circular: A = π * (d/2)²
area_m2 = math.pi * (diametro_m / 2) ** 2

# Tensão aplicada: σ = F / A (resultado em Pa) e conversão para MPa
tensao_Pa = carga_N / area_m2
tensao_MPa = tensao_Pa / 1_000_000

print(f"Diâmetro em metros: {diametro_m:.4f} m")
print(f"Área da seção:      {area_m2:.6f} m²")
print(f"Tensão aplicada:    {tensao_Pa:.2f} Pa")
print(f"Tensão aplicada:    {tensao_MPa:.2f} MPa")
```

```python
# Fator de utilização: fração da capacidade máxima em uso
fator = tensao_MPa / tensao_ruptura_MPa

# Classificação de risco
if fator < 0.50:
    classificacao = "Operação segura"
elif fator < 0.75:
    classificacao = "Atenção — verifique periodicamente"
elif fator < 0.90:
    classificacao = "Risco elevado — substituição recomendada"
else:
    classificacao = "Perigo — cabo fora de operação"

print(f"Fator de utilização: {fator:.2f}")
print(f"Classificação: {classificacao}")
```

```python
print("=== RELATÓRIO DE ANÁLISE DE CABO ===")
print(f"Diâmetro:            {diametro_mm:.2f} mm")
print(f"Carga aplicada:      {carga_N:.2f} N")
print(f"Área da seção:       {area_m2:.6f} m²")
print(f"Tensão aplicada:     {tensao_MPa:.2f} MPa")
print(f"Fator de utilização: {fator:.2f}")
print(f"Classificação:       {classificacao}")
print("=====================================")
```

## Interpretação do resultado

Sim, o cabo está em condição segura de operação. A tensão aplicada é de cerca de 75 MPa, contra 400 MPa de ruptura, o que dá um fator de utilização de aproximadamente 0,19: o cabo usa menos de um quinto da sua capacidade, ou seja, há um coeficiente de segurança em torno de 5,3.

Ao engenheiro responsável eu recomendaria manter a inspeção visual periódica (fios rompidos, corrosão, desgaste), já que o cálculo considera apenas carga estática. Em içamento real existem cargas dinâmicas (partidas bruscas, oscilações) que aumentam a tensão, e a margem atual comporta isso com folga. Se a carga ou o uso mudarem, o cálculo deve ser refeito.
