O programa simula o resfriamento de um reator minuto a minuto com um laço `while`, que só termina quando a temperatura chega ao limite seguro. A taxa de queda muda conforme a faixa de temperatura (o resfriamento fica mais lento à medida que esfria). Ao final, mostra o tempo total, a temperatura final e quantos minutos foram gastos em cada faixa.

```python
# Dados de entrada
temperatura_inicial = 850.0  # temperatura inicial do reator (°C)
temperatura_segura = 120.0   # limite de liberação do equipamento (°C)
taxa_resfriamento = 35.0     # queda de temperatura por minuto (°C/min), atualizada no loop
tempo_minuto = 0             # contador de tempo (min)

temperatura_atual = temperatura_inicial  # temperatura ao longo da simulação (°C)

# Contadores de minutos por faixa
minutos_faixa_alta = 0    # acima de 500 °C
minutos_faixa_media = 0   # entre 300 e 500 °C
minutos_faixa_baixa = 0   # entre 120 e 300 °C
```

```python
while temperatura_atual > temperatura_segura:
    # Define a taxa e o contador da faixa atual
    if temperatura_atual > 500:
        taxa_resfriamento = 35.0
        minutos_faixa_alta += 1
    elif temperatura_atual >= 300:
        taxa_resfriamento = 22.0
        minutos_faixa_media += 1
    else:
        taxa_resfriamento = 10.0
        minutos_faixa_baixa += 1

    temperatura_atual -= taxa_resfriamento
    tempo_minuto += 1

    print(f"Minuto {tempo_minuto:02d} | Temperatura: {temperatura_atual:.1f} °C | "
          f"Taxa aplicada: {taxa_resfriamento:.1f} °C/min")
```

```python
print("=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:     {temperatura_inicial:.1f} °C")
print(f"Temperatura final:       {temperatura_atual:.1f} °C")
print(f"Tempo total:             {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:    {minutos_faixa_alta} minutos")
print(f"Tempo na faixa 300-500:  {minutos_faixa_media} minutos")
print(f"Tempo na faixa 120-300:  {minutos_faixa_baixa} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

## Interpretação do resultado

O processo passou mais tempo na faixa de **120 a 300 °C**, com 16 minutos, contra 10 minutos em cada uma das outras duas faixas, mesmo cobrindo a menor variação de temperatura (180 °C). O motivo é a taxa de apenas 10 °C/min nessa faixa.

Fisicamente, a troca de calor é proporcional à diferença de temperatura entre o reator e o ambiente (lei de resfriamento de Newton). Quanto mais o reator esfria, menor é essa diferença, então o fluxo de calor diminui e a queda por minuto fica cada vez mais lenta.

Se a taxa fosse constante em 35 °C/min, o resfriamento levaria cerca de 21 minutos (730 °C ÷ 35), contra os 36 minutos da simulação. O modelo constante subestimaria o tempo em mais de 15 minutos e faria a equipe liberar o equipamento antes da hora, com risco de segurança. Ele também terminaria abaixo do limite (115 °C), enquanto o modelo com taxas por faixa chega exatamente a 120 °C.
