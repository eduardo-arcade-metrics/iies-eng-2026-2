## Descrição do problema

Este programa irá analisar o consumo de combustível de um caminhão em uma rota dividida em cinco trechos. Para cada trecho, será considerada a distância e a velocidade média para calcular o consumo de diesel e o tempo de viagem. Ao final, será apresentado o consumo, o tempo e o custo total da rota.


```python
# Dados de entrada

extensoes_km = [42, 18, 65, 30, 25]       # Distância de cada trecho em km
velocidades_kmh = [90, 40, 110, 60, 80]   # Velocidade média de cada trecho em km/h
preco_litro = 6.89                          # Preço do litro do diesel em reais

# Variáveis acumuladoras
consumo_total = 0
tempo_total = 0
```

```python
# Percorre todos os trechos da rota
for i in range(len(extensoes_km)):

    extensao = extensoes_km[i]
    velocidade = velocidades_kmh[i]

    # Determina o consumo de acordo com a velocidade
    if velocidade <= 50:
        consumo_100km = 18.0
    elif velocidade <= 80:
        consumo_100km = 14.5
    elif velocidade <= 100:
        consumo_100km = 12.0
    else:
        consumo_100km = 15.5

    # Calcula o consumo e o tempo do trecho
    consumo_trecho = (extensao * consumo_100km) / 100
    tempo_trecho = extensao / velocidade

    # Acumula os resultados
    consumo_total += consumo_trecho
    tempo_total += tempo_trecho

    # Exibe os dados de cada trecho
    print(f"Trecho {i + 1}: {extensao} km a {velocidade} km/h | "
          f"Consumo: {consumo_100km:.1f} L/100km | "
          f"Gasto: {consumo_trecho:.2f} L | "
          f"Tempo: {tempo_trecho:.2f} h")
```

```python
# Cálculo do custo total da rota
custo_total = consumo_total * preco_litro
```

```python
# Cálculo da distância total
distancia_total = sum(extensoes_km)

# Relatório final
print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total:.2f} km")
print(f"Tempo total:         {tempo_total:.2f} h")
print(f"Consumo total:       {consumo_total:.2f} litros")
print(f"Custo total:         R$ {custo_total:.2f}")
print("=========================")
```

## Interpretação do resultado

O trecho 3 apresentou o maior consumo de combustível, com aproximadamente 10,07 litros, pois além de ser o trecho mais longo da rota, com 65 km, o caminhão trafega a 110 km/h.

A velocidade mais alta nem sempre representa maior economia, pois acima de 100 km/h o consumo do caminhão aumenta. Nesse caso, o consumo passa para 15,5 L/100 km, enquanto entre 81 e 100 km/h é de 12,0 L/100 km.

Assim, manter o veículo em uma faixa de velocidade adequada pode contribuir para reduzir o consumo de combustível e o custo da viagem.
