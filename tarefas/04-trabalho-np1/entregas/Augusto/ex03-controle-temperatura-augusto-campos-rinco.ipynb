## Descrição do problema

Este programa irá simular o resfriamento de um equipamento industrial a partir de uma temperatura inicial de 850 °C. A taxa de resfriamento será alterada conforme a faixa de temperatura, e o processo continuará até atingir uma condição segura de operação. Ao final, será apresentado o tempo total e o tempo gasto em cada faixa de temperatura.

```python
# Dados de entrada

temperatura_inicial = 850.0    # Temperatura inicial do reator em °C
temperatura_segura = 120.0     # Temperatura considerada segura em °C
taxa_resfriamento = 35.0       # Taxa inicial de resfriamento em °C/min
tempo_minuto = 0               # Contador do tempo em minutos

# Temperatura atual no início do processo
temperatura_atual = temperatura_inicial

# Contadores de tempo em cada faixa de temperatura
tempo_acima_500 = 0
tempo_300_500 = 0
tempo_120_300 = 0
```

```python
# Loop de resfriamento
while temperatura_atual > temperatura_segura:

    # Determina a taxa de resfriamento de acordo com a temperatura
    if temperatura_atual > 500:
        taxa_resfriamento = 35.0
        tempo_acima_500 += 1

    elif temperatura_atual > 300:
        taxa_resfriamento = 22.0
        tempo_300_500 += 1

    else:
        taxa_resfriamento = 10.0
        tempo_120_300 += 1

    # Atualiza a temperatura
    temperatura_atual -= taxa_resfriamento

    # Incrementa o tempo
    tempo_minuto += 1

    # Exibe o estado atual do processo
    print(f"Minuto {tempo_minuto:02d} | "
          f"Temperatura: {temperatura_atual:.1f} °C | "
          f"Taxa aplicada: {taxa_resfriamento:.1f} °C/min")
```

```python
# Relatório final do processo

print("\n=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:    {temperatura_inicial:.1f} °C")
print(f"Temperatura final:      {temperatura_atual:.1f} °C")
print(f"Tempo total:            {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:   {tempo_acima_500} minutos")
print(f"Tempo na faixa 300-500: {tempo_300_500} minutos")
print(f"Tempo na faixa 120-300: {tempo_120_300} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

## Interpretação do resultado

O processo passou mais tempo na faixa entre 120 °C e 300 °C, permanecendo nela durante 16 minutos. Isso ocorre porque, conforme a temperatura diminui, o gradiente térmico também diminui e o processo de resfriamento se torna mais lento.

Nessa faixa, a taxa utilizada é de apenas 10 °C/min, enquanto nas temperaturas acima de 500 °C a taxa é de 35 °C/min. Por isso, mesmo com uma diferença menor de temperatura, são necessários mais minutos para completar essa etapa.

Se a taxa permanecesse constante em 35 °C/min durante todo o processo, o equipamento atingiria a temperatura segura em menos tempo. Porém, esse comportamento não representaria o modelo de resfriamento proposto, no qual a taxa diminui conforme a temperatura do equipamento se aproxima da condição segura.
