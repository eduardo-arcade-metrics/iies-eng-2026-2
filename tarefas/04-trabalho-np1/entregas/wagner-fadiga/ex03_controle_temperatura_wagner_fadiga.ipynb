# Exercício 03 — Controle de temperatura de reator

## Descrição

Neste exercício será realizado o controle da temperatura de um reator.
O programa irá verificar continuamente a temperatura até que ela esteja dentro da faixa adequada.
Será utilizado um laço `while` para repetir as verificações.

```python
# Dados iniciais do reator
temperatura = 70
temperatura_minima = 60
temperatura_maxima = 80
```

```python
# Controle da temperatura
while temperatura < temperatura_minima or temperatura > temperatura_maxima:
    print(f"Temperatura atual: {temperatura:.2f} °C")

    if temperatura < temperatura_minima:
        temperatura += 5
    elif temperatura > temperatura_maxima:
        temperatura -= 5

print(f"Temperatura final: {temperatura:.2f} °C")
```

```python
# Relatório final
print("=== CONTROLE DE TEMPERATURA DO REATOR ===")
print(f"Temperatura inicial: {70:.2f} °C")
print(f"Faixa adequada: {temperatura_minima:.2f} °C a {temperatura_maxima:.2f} °C")
print(f"Temperatura final: {temperatura:.2f} °C")
print("Status: Temperatura dentro da faixa adequada.")
print("==========================================")
```

## Interpretação do resultado

O programa verificou a temperatura do reator utilizando um laço `while`.
A temperatura foi ajustada até atingir a faixa adequada de operação.
Ao final, a temperatura ficou dentro dos limites estabelecidos para o reator.
