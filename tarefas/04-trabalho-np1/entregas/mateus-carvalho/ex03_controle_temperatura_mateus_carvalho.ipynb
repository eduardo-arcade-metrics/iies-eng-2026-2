# Exercício 03 — Controle de temperatura em processo de resfriamento industrial

Este programa simula o processo de resfriamento controlado de um reator industrial até atingir uma temperatura segura.
Utilizando a estrutura de repetição `while`, o algoritmo monitora a temperatura minuto a minuto, aplicando taxas de resfriamento decrescentes conforme o equipamento esfria.
Ao final, o sistema apresenta o tempo total de resfriamento e o tempo de permanência em cada faixa operacional.


```python
# Inicialização das variáveis de controle fornecidas pelo enunciado
temperatura_inicial = 850.0    # Temperatura de partida do reator (°C)
temperatura_segura = 120.0     # Limite seguro para liberação do equipamento (°C)
temperatura_atual = 850.0      # Variável que será atualizada no processo (°C)
tempo_minuto = 0               # Contador geral de tempo (minutos)

# Contadores específicos para mensurar o tempo gasto em cada faixa térmica
tempo_faixa_alta = 0           # Minutos com temperatura acima de 500°C
tempo_faixa_media = 0          # Minutos com temperatura entre 300°C e 500°C
tempo_faixa_baixa = 0          # Minutos com temperatura entre 120°C e 300°C
```

```python
# Loop condicional executado enquanto a temperatura atual for maior que o limite seguro
while temperatura_atual > temperatura_segura:
    # Incrementa o contador de tempo no início do intervalo correspondente
    tempo_minuto += 1

    # Estrutura condicional para determinar a taxa de resfriamento e atualizar os contadores de faixa
    if temperatura_atual > 500:
        taxa_atual = 35.0
        tempo_faixa_alta += 1
    elif temperatura_atual >= 300:
        taxa_atual = 22.0
        tempo_faixa_media += 1
    else:
        taxa_atual = 10.0
        tempo_faixa_baixa += 1

    # Aplicação da taxa para atualização da temperatura atual do reator
    temperatura_atual -= taxa_atual

    # Exibição do monitoramento em tempo real minuto a minuto conforme padrão solicitado
    print(f"Minuto {tempo_minuto:02d} | Temperatura: {temperatura_atual:.1f} °C | Taxa aplicada: {taxa_atual:.1f} °C/min")
```

```python
# Exibição do relatório de encerramento consolidado utilizando f-strings
print("\n=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:   {temperatura_inicial:.1f} °C")
print(f"Temperatura final:     {temperatura_atual:.1f} °C")
print(f"Tempo total:           {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:  {tempo_faixa_alta} minutos")
print(f"Tempo na faixa 300-500: {tempo_faixa_media} minutos")
print(f"Tempo na faixa 120-300: {tempo_faixa_baixa} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

### Interpretação do Resultado
O processo passou a maior parte do tempo na **faixa baixa (entre 120°C e 300°C)**, necessitando de um total de **17 minutos** de resfriamento apenas nessa etapa. Fisicamente isso ocorre porque, à medida que a temperatura do reator se aproxima da temperatura ambiente, a diferença térmica (gradiente térmico) diminui, tornando a perda de calor por convecção e radiação naturalmente mais lenta. Caso a taxa de resfriamento fosse mantida constante em 35 °C/min durante todo o processo, o reator alcançaria o limite seguro muito mais rápido (em cerca de 21 minutos no total contra os 37 minutos reais), porém esse choque térmico acelerado poderia causar tensões mecânicas severas ou trincas estruturais no material do equipamento.
