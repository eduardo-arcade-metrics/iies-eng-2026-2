# Exercício 02 — Consumo de combustível em rota de transporte
**Estudante:** Mateus Carvalho



Este programa realiza a estimativa do consumo de diesel e tempo de viagem de um caminhão ao longo de uma rota dividida em cinco trechos com velocidades variáveis.
O algoritmo percorre cada trecho aplicando regras específicas de consumo conforme a velocidade de operação e acumula os valores totais de combustível gasto e tempo de trânsito.
Ao final, calcula o custo financeiro total com base no preço por litro do combustível.


```python
# Listas com os dados de cada trecho fornecidos no enunciado
extensoes_km = [42, 18, 65, 30, 25]        # Distância de cada um dos 5 trechos (km)
velocidades_kmh = [90, 40, 110, 60, 80]    # Velocidade média em cada trecho (km/h)

# Preço de mercado do litro de diesel para o cálculo financeiro
preco_litro = 6.89                         # Preço do combustível em Reais (R$)
```

```python
# Inicialização das variáveis acumuladoras para os totais da rota
consumo_total_litros = 0.0
tempo_total_horas = 0.0
distancia_total_km = 0.0

# Loop utilizando range() para percorrer as listas com base no índice dos trechos
for i in range(len(extensoes_km)):
    # Resgate das informações do trecho atual usando o índice i
    km = extensoes_km[i]
    v_kmh = velocidades_kmh[i]

    # Estrutura condicional para determinar o consumo padrão (L/100km) conforme a velocidade
    if v_kmh <= 50:
        consumo_100km = 18.0
    elif v_kmh <= 80:
        consumo_100km = 14.5
    elif v_kmh <= 100:
        consumo_100km = 12.0
    else:
        consumo_100km = 15.5

    # Cálculos matemáticos específicos do trecho utilizando as fórmulas fornecidas
    gasto_trecho = (km * consumo_100km) / 100
    tempo_trecho = km / v_kmh

    # Acumulando os valores parciais nas variáveis globais da rota
    distancia_total_km += km
    consumo_total_litros += gasto_trecho
    tempo_total_horas += tempo_trecho

    # Exibição dos dados do trecho atual conforme o formato solicitado
    print(f"Trecho {i+1}: {km} km a {v_kmh} km/h | Consumo: {consumo_100km:.1f} L/100km | Gasto: {gasto_trecho:.2f} L | Tempo: {tempo_trecho:.2f} h")
```

```python
# Cálculo do custo financeiro total multiplicando os litros acumulados pelo preço do diesel
custo_total_reais = consumo_total_litros * preco_litro
```

```python
# Impressão do relatório estruturado utilizando f-strings com duas casas decimais
print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total_km:.2f} km")
print(f"Tempo total:         {tempo_total_horas:.2f} h")
print(f"Consumo total:       {consumo_total_litros:.2f} litros")
print(f"Custo total:         R$ {custo_total_reais:.2f}")
print("=========================")
```

### Interpretação do Resultado
O trecho mais caro em termos de consumo absoluto foi o **Trecho 3** (65 km a 110 km/h), que demandou um gasto total de **10,07 litros** de óleo diesel. A velocidade mais alta nem sempre é a mais econômica porque, acima de um limite operacional (geralmente próximo de 80 km/h a 90 km/h), o arrasto aerodinâmico (a força de resistência do ar) cresce exponencialmente. Isso obriga o motor do caminhão a injetar muito mais combustível para vencer a barreira do ar e manter a velocidade elevada, reduzindo a eficiência e aumentando o consumo por quilômetro rodado.
