Este programa analisa a condição de operação de um cabo de aço submetido a uma determinada carga.
Primeiro será calculada a área da seção transversal e a tensão aplicada no cabo.
Depois, será calculado o fator de utilização para classificar o nível de risco da operação.

```python
diametro_mm = 12          # diâmetro do cabo em milímetros
carga_N = 8500            # carga aplicada em Newtons
tensao_ruptura_MPa = 400  # tensão de ruptura do aço em MPa
```

```python
import math

diametro_m = diametro_mm / 1000

area_m2 = math.pi * (diametro_m / 2) ** 2

tensao_Pa = carga_N / area_m2

tensao_MPa = tensao_Pa / 1_000_000

print(f"Área da seção: {area_m2:.6f} m²")
print(f"Tensão aplicada: {tensao_MPa:.2f} MPa")
```

```python
fator_utilizacao = tensao_MPa / tensao_ruptura_MPa

if fator_utilizacao < 0.50:
    classificacao = "Operação segura"
elif fator_utilizacao < 0.75:
    classificacao = "Atenção — verificar periodicamente"
elif fator_utilizacao < 0.90:
    classificacao = "Risco elevado — substituição recomendada"
else:
    classificacao = "Perigo — cabo fora de operação"

print(f"Fator de utilização: {fator_utilizacao:.2f}")
print(f"Classificação: {classificacao}")
```

```python
print("=== RELATÓRIO DE ANÁLISE DE CABO ===")
print(f"Diâmetro:            {diametro_mm:.2f} mm")
print(f"Carga aplicada:      {carga_N:.2f} N")
print(f"Área da seção:       {area_m2:.6f} m²")
print(f"Tensão aplicada:     {tensao_MPa:.2f} MPa")
print(f"Fator de utilização: {fator_utilizacao:.2f}")
print(f"Classificação:       {classificacao}")
print("=====================================")
```

## Interpretação do resultado

O cabo apresenta um fator de utilização de aproximadamente 0,19, ficando abaixo do limite de 0,50 estabelecido para operação segura. Portanto, considerando os dados fornecidos no exercício, o cabo está em condição segura de operação. Recomenda-se, ainda, que o engenheiro responsável mantenha as inspeções periódicas e verifique as condições reais do cabo antes e durante sua utilização.
