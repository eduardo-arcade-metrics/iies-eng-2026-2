# Exercício 03 — Controle de temperatura em processo de resfriamento industrial

**Nome:** Gustavo Fernando R S Leite

Este programa simula o resfriamento de um reator industrial a partir de uma temperatura inicial elevada.
A cada minuto, o programa verifica a temperatura e determina a taxa de resfriamento correspondente à faixa atual.
O processo continua até que a temperatura atinja o limite seguro, registrando o tempo e as faixas de temperatura percorridas.

```python
temperatura_inicial = 850.0    # temperatura inicial do reator em graus Celsius
temperatura_segura = 120.0     # temperatura segura para liberação em graus Celsius
taxa_resfriamento = 35.0       # taxa inicial de resfriamento em graus Celsius por minuto
tempo_minuto = 0               # contador de tempo em minutos

contador_acima_500 = 0         # minutos na faixa acima de 500 °C
contador_300_500 = 0            # minutos na faixa entre 300 °C e 500 °C
contador_120_300 = 0            # minutos na faixa entre 120 °C e 300 °C

temperatura_atual = temperatura_inicial
```

```python
while temperatura_atual > temperatura_segura:

    if temperatura_atual > 500:
        taxa_resfriamento = 35.0
        contador_acima_500 += 1

    elif temperatura_atual >= 300:
        taxa_resfriamento = 22.0
        contador_300_500 += 1

    else:
        taxa_resfriamento = 10.0
        contador_120_300 += 1

    temperatura_atual -= taxa_resfriamento
    tempo_minuto += 1

    print(
        f"Minuto {tempo_minuto:02d} | "
        f"Temperatura: {temperatura_atual:.1f} °C | "
        f"Taxa aplicada: {taxa_resfriamento:.1f} °C/min"
    )
```

```python
print("=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:   {temperatura_inicial:.1f} °C")
print(f"Temperatura final:     {temperatura_atual:.1f} °C")
print(f"Tempo total:           {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:  {contador_acima_500} minutos")
print(f"Tempo na faixa 300-500: {contador_300_500} minutos")
print(f"Tempo na faixa 120-300: {contador_120_300} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

## Interpretação do resultado

O processo passou mais tempo na faixa entre 120 °C e 300 °C, porque nessa faixa a taxa de resfriamento é menor, de apenas 10 °C por minuto. Isso representa fisicamente um resfriamento mais lento conforme a diferença de temperatura diminui.

Se a taxa de resfriamento fosse constante em 35 °C/min durante todo o processo, o reator atingiria a temperatura segura em menos tempo. Porém, esse modelo não representaria a redução natural da velocidade de resfriamento que ocorre conforme a temperatura do equipamento se aproxima da temperatura ambiente.
