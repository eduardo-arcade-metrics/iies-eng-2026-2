Este programa calcula o consumo de combustível de um caminhão durante uma rota dividida em cinco trechos.
Para cada trecho, o programa considera a distância e a velocidade média para determinar o consumo.
Ao final, serão calculados o tempo total da viagem, o consumo total de combustível e o custo da rota.

```python
extensoes_km = [42, 18, 65, 30, 25]        # distância de cada trecho em quilômetros
velocidades_kmh = [90, 40, 110, 60, 80]    # velocidade média de cada trecho em km/h
preco_litro = 6.89                          # preço de um litro de diesel em reais
```

```python
consumo_total_litros = 0
tempo_total_h = 0

for i in range(len(extensoes_km)):
    extensao_km = extensoes_km[i]
    velocidade_kmh = velocidades_kmh[i]

    if velocidade_kmh <= 50:
        consumo_L_por_100km = 18.0
    elif velocidade_kmh <= 80:
        consumo_L_por_100km = 14.5
    elif velocidade_kmh <= 100:
        consumo_L_por_100km = 12.0
    else:
        consumo_L_por_100km = 15.5

    consumo_trecho = (extensao_km * consumo_L_por_100km) / 100

    tempo_trecho_h = extensao_km / velocidade_kmh

    consumo_total_litros += consumo_trecho
    tempo_total_h += tempo_trecho_h

    print(
        f"Trecho {i + 1}: {extensao_km} km a {velocidade_kmh} km/h | "
        f"Consumo: {consumo_L_por_100km:.1f} L/100km | "
        f"Gasto: {consumo_trecho:.2f} L | "
        f"Tempo: {tempo_trecho_h:.2f} h"
    )
```

```python
custo_total = consumo_total_litros * preco_litro

print(f"Consumo total: {consumo_total_litros:.2f} litros")
print(f"Custo total: R$ {custo_total:.2f}")
```

```python
distancia_total = sum(extensoes_km)

print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total:.2f} km")
print(f"Tempo total:         {tempo_total_h:.2f} h")
print(f"Consumo total:       {consumo_total_litros:.2f} litros")
print(f"Custo total:         R$ {custo_total:.2f}")
print("=========================")
```

## Interpretação do resultado

O trecho 3 foi o que apresentou o maior consumo de combustível, com aproximadamente 10,08 litros, pois possui 65 km de extensão e uma velocidade média de 110 km/h. Nessa velocidade, o consumo aumenta para 15,5 L/100 km devido ao maior arrasto aerodinâmico.

A velocidade mais alta nem sempre é a mais econômica porque o consumo depende da faixa de velocidade do veículo. Em velocidades muito altas, o arrasto aerodinâmico aumenta e faz o caminhão consumir mais combustível por quilômetro.
