Neste exercício, vamos calcular o consumo de combustível de um caminhão que percorre uma rota dividida em cinco trechos. Para cada trecho, o programa identifica o consumo de acordo com a velocidade, calcula o combustível gasto e o tempo de viagem. Ao final, serão calculados o consumo total, o tempo total e o custo da viagem.

```python
# Dados de entrada
extensoes_km = [42, 18, 65, 30, 25]        # distância de cada trecho em km
velocidades_kmh = [90, 40, 110, 60, 80]    # velocidade média de cada trecho em km/h
preco_litro = 6.89                         # preço do litro do diesel em reais

# Variáveis para acumular os resultados
consumo_total_litros = 0
tempo_total_h = 0
```

```python
# Cálculo de cada trecho
for i in range(len(extensoes_km)):
    extensao_km = extensoes_km[i]
    velocidade_kmh = velocidades_kmh[i]

    # Define o consumo de acordo com a velocidade média
    if velocidade_kmh <= 50:
        consumo_L_por_100km = 18.0
    elif velocidade_kmh <= 80:
        consumo_L_por_100km = 14.5
    elif velocidade_kmh <= 100:
        consumo_L_por_100km = 12.0
    else:
        consumo_L_por_100km = 15.5

    # Calcula o consumo em litros no trecho
    consumo_trecho = (extensao_km * consumo_L_por_100km) / 100

    # Calcula o tempo de viagem em horas
    tempo_trecho_h = extensao_km / velocidade_kmh

    # Acumula os resultados
    consumo_total_litros += consumo_trecho
    tempo_total_h += tempo_trecho_h

    # Exibe os dados do trecho
    print(f"Trecho {i + 1}: {extensao_km} km a {velocidade_kmh} km/h | "
          f"Consumo: {consumo_L_por_100km:.1f} L/100km | "
          f"Gasto: {consumo_trecho:.2f} L | "
          f"Tempo: {tempo_trecho_h:.2f} h")
```

```python
# Cálculo do custo total da rota
custo_total = consumo_total_litros * preco_litro

print(f"Custo total de combustível: R$ {custo_total:.2f}")
```

```python
# Relatório final da rota
distancia_total = sum(extensoes_km)

print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total:.2f} km")
print(f"Tempo total:         {tempo_total_h:.2f} h")
print(f"Consumo total:       {consumo_total_litros:.2f} litros")
print(f"Custo total:         R$ {custo_total:.2f}")
print("=========================")
```

## Interpretação do resultado

O trecho 3 foi o que apresentou o maior consumo de combustível, com aproximadamente **10,08 litros**, pois possui 65 km de extensão e velocidade média de 110 km/h, faixa em que o consumo é de 15,5 L/100 km.

A velocidade mais alta nem sempre é a mais econômica porque, acima de 100 km/h, o arrasto aerodinâmico aumenta e faz o consumo subir. Além disso, velocidades muito baixas também apresentam consumo maior, conforme a tabela do exercício.

Assim, existe uma faixa de velocidade mais econômica, e o consumo depende tanto da velocidade quanto da distância percorrida em cada trecho.
