## Descrição

Neste exercício será analisado um cabo de aço que suporta uma determinada carga.  
O programa irá calcular a área da seção transversal, a tensão aplicada e o fator de utilização.  
Ao final, o cabo será classificado de acordo com o nível de risco operacional.

```python
# Dados do cabo
diametro_mm = 12        # diâmetro do cabo em milímetros
carga_N = 8500          # força aplicada em Newtons
tensao_ruptura_MPa = 400  # tensão de ruptura do material em MPa
```

```python
import math

# Conversão do diâmetro de milímetros para metros
diametro_m = diametro_mm / 1000

# Cálculo do raio
raio_m = diametro_m / 2

# Cálculo da área da seção transversal
area_m2 = math.pi * (raio_m ** 2)

# Cálculo da tensão aplicada em Pascal
tensao_Pa = carga_N / area_m2

# Conversão da tensão de Pascal para MPa
tensao_MPa = tensao_Pa / 1_000_000

# Exibição dos resultados intermediários
print(f"Área da seção transversal: {area_m2:.6f} m²")
print(f"Tensão aplicada: {tensao_MPa:.2f} MPa")
```

```python
# Cálculo do fator de utilização
fator_utilizacao = tensao_MPa / tensao_ruptura_MPa

# Classificação do cabo
if fator_utilizacao < 0.50:
    classificacao = "Operação segura"
elif fator_utilizacao < 0.75:
    classificacao = "Atenção — verificar periodicamente"
elif fator_utilizacao < 0.90:
    classificacao = "Risco elevado — substituição recomendada"
else:
    classificacao = "Perigo — cabo fora de operação"
```

```python
print("=== RELATÓRIO DE ANÁLISE DE CABO ===")
print(f"Diâmetro:            {diametro_mm:.2f} mm")
print(f"Carga aplicada:      {carga_N:.2f} N")
print(f"Área da seção:       {area_m2:.6f} m²")
print(f"Tensão aplicada:     {tensao_MPa:.2f} MPa")
print(f"Fator de utilização: {fator_utilizacao:.2f}")
print(f"Classificação:       {classificacao}")
print("=====================================")
```

## Interpretação do resultado

O cabo está em condição segura de operação, pois o fator de utilização ficou abaixo de 0,50.  
A tensão aplicada está bem abaixo da tensão de ruptura do material.  
Recomenda-se manter o acompanhamento periódico do cabo e realizar inspeções para garantir sua segurança durante a operação.
