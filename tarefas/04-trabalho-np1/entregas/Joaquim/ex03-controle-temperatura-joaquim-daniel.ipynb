# Exercício 03 — Controle de temperatura em processo de resfriamento industrial

Neste exercício, vamos simular o resfriamento de um equipamento industrial a partir de uma temperatura inicial de 850 °C. O programa utiliza um loop `while` para acompanhar a temperatura minuto a minuto, aplicando diferentes taxas de resfriamento conforme a faixa de temperatura. Ao final, serão apresentados o tempo total, a temperatura final e o tempo gasto em cada faixa.

```python
# Dados de entrada
temperatura_inicial = 850.0    # temperatura inicial do reator em graus Celsius
temperatura_segura = 120.0     # limite de liberação do equipamento em graus Celsius
taxa_resfriamento = 35.0       # queda de temperatura por minuto em graus Celsius
tempo_minuto = 0               # contador de tempo em minutos

# Contadores para cada faixa de temperatura
contador_acima_500 = 0
contador_300_500 = 0
contador_120_300 = 0

# Variável que será atualizada durante o resfriamento
temperatura = temperatura_inicial
```

```python
# Loop de resfriamento
while temperatura > temperatura_segura:
    # Determina a taxa de resfriamento e incrementa o contador da faixa atual
    if temperatura > 500:
        taxa_atual = 35.0
        contador_acima_500 += 1
    elif temperatura >= 300:
        taxa_atual = 22.0
        contador_300_500 += 1
    else:
        taxa_atual = 10.0
        contador_120_300 += 1

    # Atualiza a temperatura e o tempo
    temperatura -= taxa_atual
    tempo_minuto += 1

    # Registra o estado atual do processo
    print(f"Minuto {tempo_minuto:02d} | Temperatura: {temperatura:.1f} °C | "
          f"Taxa aplicada: {taxa_atual:.1f} °C/min")
```

```python
# Relatório final
print("=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:   {temperatura_inicial:.1f} °C")
print(f"Temperatura final:     {temperatura:.1f} °C")
print(f"Tempo total:           {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:  {contador_acima_500} minutos")
print(f"Tempo na faixa 300-500: {contador_300_500} minutos")
print(f"Tempo na faixa 120-300: {contador_120_300} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

## Interpretação do resultado

O processo passou mais tempo na faixa entre 120 °C e 300 °C, pois nessa faixa a taxa de resfriamento é menor, de apenas 10 °C por minuto. Isso acontece fisicamente porque, conforme a temperatura do equipamento se aproxima da temperatura ambiente, o gradiente térmico diminui e o processo de resfriamento fica mais lento.

Se a taxa de resfriamento fosse constante em 35 °C/min durante todo o processo, o equipamento atingiria a temperatura segura muito mais rapidamente. Nesse caso, a temperatura cairia de forma uniforme, sem considerar a redução da velocidade de resfriamento nas temperaturas mais baixas.

O programa mantém o comportamento por intervalos discretos, portanto a temperatura final pode ficar abaixo do limite de 120 °C. Isso é esperado conforme o enunciado.
