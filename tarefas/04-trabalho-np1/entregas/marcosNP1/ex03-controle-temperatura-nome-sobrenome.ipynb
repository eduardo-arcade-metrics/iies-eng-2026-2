## Descrição do problema

Este notebook simula o resfriamento de um reator industrial, partindo de uma temperatura inicial de 850 °C até atingir a temperatura de segurança de 120 °C. A taxa de resfriamento não é constante: ela diminui conforme a temperatura do equipamento cai, refletindo a redução do gradiente térmico. O programa registra a temperatura minuto a minuto usando um loop `while` e, ao final, informa o tempo total de resfriamento e quanto tempo o processo permaneceu em cada faixa de temperatura.

```python
# Célula 3 — Dados de entrada

temperatura_atual = 850.0      # temperatura inicial do reator, em graus Celsius (será atualizada a cada minuto)
temperatura_segura = 120.0     # limite de liberação do equipamento, em graus Celsius
tempo_minuto = 0               # contador de tempo, em minutos

# Contadores de minutos passados em cada faixa de temperatura
cont_acima_500 = 0     # minutos com temperatura acima de 500 °C
cont_300_500 = 0       # minutos com temperatura entre 300 e 500 °C
cont_120_300 = 0       # minutos com temperatura entre 120 e 300 °C
```

```python
# Célula 4 — Loop de resfriamento

while temperatura_atual > temperatura_segura:
    # Determina a taxa de resfriamento de acordo com a faixa de temperatura atual
    # e incrementa o contador da faixa correspondente
    if temperatura_atual > 500:
        taxa_resfriamento = 35.0
        cont_acima_500 += 1
    elif temperatura_atual >= 300:
        taxa_resfriamento = 22.0
        cont_300_500 += 1
    else:
        taxa_resfriamento = 10.0
        cont_120_300 += 1

    # Atualiza a temperatura e o contador de tempo
    temperatura_atual -= taxa_resfriamento
    tempo_minuto += 1

    # Imprime o estado do minuto atual
    print(f"Minuto {tempo_minuto:02d} | Temperatura: {temperatura_atual:.1f} °C | "
          f"Taxa aplicada: {taxa_resfriamento:.1f} °C/min")
```

```python
# Célula 5 — Saída final

print("=== RELATÓRIO DE RESFRIAMENTO ===")
print(f"Temperatura inicial:   850.0 °C")
print(f"Temperatura final:     {temperatura_atual:.1f} °C")
print(f"Tempo total:           {tempo_minuto} minutos")
print(f"Tempo na faixa > 500:  {cont_acima_500} minutos")
print(f"Tempo na faixa 300-500: {cont_300_500} minutos")
print(f"Tempo na faixa 120-300: {cont_120_300} minutos")
print("Status: Equipamento liberado para operação")
print("=================================")
```

## Interpretação do resultado

O processo passou mais tempo na faixa entre 120 e 300 °C, pois nessa faixa a taxa de resfriamento é a mais baixa (10 °C/min), fazendo com que cada grau perdido exija mais minutos. Isso acontece fisicamente porque o gradiente térmico entre o equipamento e o ambiente diminui à medida que a temperatura cai, reduzindo a taxa de troca de calor por convecção e radiação — quanto mais próxima a temperatura do reator fica da temperatura ambiente, mais lenta é a perda de calor. Se a taxa de resfriamento fosse constante em 35 °C/min durante todo o processo, o reator atingiria os 120 °C muito mais rápido, em poucos minutos, mas esse cenário não reproduziria o comportamento físico real do sistema, que naturalmente desacelera conforme a diferença de temperatura diminui. Um resfriamento constante e agressivo também poderia causar choques térmicos indesejados no equipamento em faixas de temperatura mais baixas.
