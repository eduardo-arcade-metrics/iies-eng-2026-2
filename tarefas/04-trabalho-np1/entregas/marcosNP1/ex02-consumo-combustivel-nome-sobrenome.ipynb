## Descrição do problema

Este notebook calcula o consumo de combustível de um caminhão que percorre uma rota dividida em 5 trechos, cada um com sua própria extensão e velocidade média. Para cada trecho, o consumo em litros por 100 km depende da faixa de velocidade, de acordo com uma tabela fornecida. O programa acumula o consumo total e o tempo total da rota e, ao final, calcula o custo total considerando o preço do diesel.

```python
# Célula 3 — Dados de entrada

extensoes_km = [42, 18, 65, 30, 25]        # distância de cada trecho, em km
velocidades_kmh = [90, 40, 110, 60, 80]    # velocidade média de cada trecho, em km/h

preco_litro = 6.89   # preço do litro do diesel, em reais
```

```python
# Célula 4 — Loop de cálculo por trecho

consumo_total_litros = 0   # acumulador do consumo total da rota, em litros
tempo_total_h = 0          # acumulador do tempo total da rota, em horas

for i in range(len(extensoes_km)):
    extensao_km = extensoes_km[i]        # extensão do trecho atual
    velocidade_kmh = velocidades_kmh[i]  # velocidade média do trecho atual

    # Determina o consumo em L/100km de acordo com a faixa de velocidade
    if velocidade_kmh <= 50:
        consumo_L_por_100km = 18.0
    elif velocidade_kmh <= 80:
        consumo_L_por_100km = 14.5
    elif velocidade_kmh <= 100:
        consumo_L_por_100km = 12.0
    else:
        consumo_L_por_100km = 15.5

    # Calcula o consumo em litros do trecho
    consumo_trecho = (extensao_km * consumo_L_por_100km) / 100

    # Calcula o tempo de viagem do trecho, em horas
    tempo_trecho_h = extensao_km / velocidade_kmh

    # Acumula os totais da rota
    consumo_total_litros += consumo_trecho
    tempo_total_h += tempo_trecho_h

    # Imprime os dados do trecho atual
    print(f"Trecho {i+1}: {extensao_km} km a {velocidade_kmh} km/h | "
          f"Consumo: {consumo_L_por_100km:.1f} L/100km | "
          f"Gasto: {consumo_trecho:.2f} L | "
          f"Tempo: {tempo_trecho_h:.2f} h")
```

```python
# Célula 5 — Cálculo do custo

custo_total = consumo_total_litros * preco_litro   # custo total da rota, em reais
```

```python
# Célula 6 — Saída final

distancia_total_km = sum(extensoes_km)  # distância total percorrida na rota

print("=== RELATÓRIO DE ROTA ===")
print(f"Distância total:     {distancia_total_km:.2f} km")
print(f"Tempo total:         {tempo_total_h:.2f} h")
print(f"Consumo total:       {consumo_total_litros:.2f} litros")
print(f"Custo total:         R$ {custo_total:.2f}")
print("=========================")
```

## Interpretação do resultado

O trecho mais caro em termos de consumo foi o trecho 3 (65 km a 110 km/h), pois é o mais longo da rota e, além disso, tem velocidade acima de 100 km/h, faixa em que o consumo por 100 km volta a subir por causa do arrasto aerodinâmico. A velocidade mais alta nem sempre é a mais econômica porque, embora reduza o tempo de viagem, o aumento do arrasto do ar em altas velocidades faz o motor gastar mais combustível por quilômetro rodado. Já em velocidades muito baixas, o motor também é ineficiente por trabalhar próximo da marcha lenta, o que explica por que existe uma faixa intermediária de velocidade (entre 51 e 100 km/h) mais econômica.
