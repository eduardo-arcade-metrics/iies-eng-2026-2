## Descrição

Neste exercício será analisada a capacidade de um cabo de aço submetido a uma determinada carga. O programa irá calcular a área da seção transversal, a tensão aplicada e o fator de utilização. Ao final, o cabo será classificado de acordo com o nível de risco da carga aplicada.

```python
# Dados de entrada

# Diâmetro do cabo em milímetros
diametro_mm = 12

# Carga aplicada em Newtons
carga_N = 8500

# Tensão de ruptura do aço em MPa
tensao_ruptura_MPa = 400
```

```python
import math

# Conversão do diâmetro de milímetros para metros
diametro_m = diametro_mm / 1000

# Cálculo da área da seção transversal circular
area_m2 = math.pi * (diametro_m / 2) ** 2

# Cálculo da tensão aplicada em Pascal
tensao_Pa = carga_N / area_m2

# Conversão da tensão de Pascal para MPa
tensao_MPa = tensao_Pa / 1_000_000

print(f"Área da seção transversal: {area_m2:.6f} m²")
print(f"Tensão aplicada: {tensao_MPa:.2f} MPa")
```

```python
# Cálculo do fator de utilização
fator_utilizacao = tensao_MPa / tensao_ruptura_MPa

# Classificação do risco
if fator_utilizacao < 0.50:
    classificacao = "Operação segura"
elif fator_utilizacao < 0.75:
    classificacao = "Atenção — verificar periodicamente"
elif fator_utilizacao < 0.90:
    classificacao = "Risco elevado — substituição recomendada"
else:
    classificacao = "Perigo — cabo fora de operação"
```

```python
print("=== RELATÓRIO DE ANÁLISE DE CABO ===")
print(f"Diâmetro:            {diametro_mm:.2f} mm")
print(f"Carga aplicada:      {carga_N:.2f} N")
print(f"Área da seção:       {area_m2:.6f} m²")
print(f"Tensão aplicada:     {tensao_MPa:.2f} MPa")
print(f"Fator de utilização: {fator_utilizacao:.2f}")
print(f"Classificação:       {classificacao}")
print("=====================================")
```

## Interpretação do resultado

O cabo apresenta um fator de utilização baixo em relação à sua tensão de ruptura, portanto está dentro da faixa classificada como operação segura.

Mesmo estando nessa condição, é importante realizar inspeções periódicas para verificar o estado físico do cabo e identificar possíveis desgastes ou danos.

Recomenda-se que o engenheiro responsável mantenha o acompanhamento das condições de uso e das cargas aplicadas, garantindo que os limites de segurança sejam respeitados.
