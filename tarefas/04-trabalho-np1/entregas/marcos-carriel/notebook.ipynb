## Descrição

Neste exercício será analisada a capacidade de um cabo de aço submetido a uma determinada carga.
O programa irá calcular a área da seção transversal, a tensão aplicada e o fator de utilização.
Ao final, o cabo será classificado de acordo com o nível de risco da carga aplicada.

```python
# Diâmetro do cabo em milímetros
diametro_mm = 12

# Carga aplicada em Newtons
carga_N = 8500

# Tensão de ruptura do aço em MPa
tensao_ruptura_MPa = 400

#DESCRIÇÃO DOS DADOS PARA REALIZAR OS CALCULOS
```

```python
import math

# Conversão do diâmetro de milímetros para metros
diametro_m = diametro_mm / 1000

# Cálculo da área da seção transversal circular
area_m2 = math.pi * (diametro_m / 2) ** 2

# Cálculo da tensão aplicada em Pascal
tensao_Pa = carga_N / area_m2

# Conversão da tensão de Pascal para MPa
tensao_MPa = tensao_Pa / 1_000_000

print(f"Área da seção transversal: {area_m2:.6f} m²")
print(f"Tensão aplicada: {tensao_MPa:.2f} MPa")

#nesse codigo é aplicado a conversão do diametro (diametro_mm) de milimetros para metros para aplicação da formula
# na seguencia  realiza o calculo de tenção aplicada com a conta de (diamentro_m/2)**2
# com esse resultado ele calcula a tenção aplicada em pascal com (carga_N / area_m2)
# e realiza a conversao para MPA com a conta ( tensao_Pa / 1_000_000)
#na seguencia ele imprime a rea da seção transversal : {area_m2:.6f} m²")
# e imprime a tenão aplicada {tensao_MPa:.2f} MPa")
```

```python
# Cálculo do fator de utilização
fator_utilizacao = tensao_MPa / tensao_ruptura_MPa

# Classificação do risco
if fator_utilizacao < 0.50:
    classificacao = "Operação segura"
elif fator_utilizacao < 0.75:
    classificacao = "Atenção — verificar periodicamente"
elif fator_utilizacao < 0.90:
    classificacao = "Risco elevado — substituição recomendada"
else:
    classificacao = "Perigo — cabo fora de operação"

    #aqui ele utiliza o if e elif para classificar os dados coletados de acordo com os resultados das formulas
```

```python
print("=== RELATÓRIO DE ANÁLISE DE CABO ===")
print(f"Diâmetro:            {diametro_mm:.2f} mm")
print(f"Carga aplicada:      {carga_N:.2f} N")
print(f"Área da seção:       {area_m2:.6f} m²")
print(f"Tensão aplicada:     {tensao_MPa:.2f} MPa")
print(f"Fator de utilização: {fator_utilizacao:.2f}")
print(f"Classificação:       {classificacao}")
print("=====================================")

#de acordo com as classificações ele imprime o resultado (classificação ) junto com os outros dados coletados
```
